## Triển khai
1. Giải nén `NgocMy_TikTokAds_Streamlit.zip`.
2. Đưa toàn bộ file vào thư mục gốc của repository GitHub.
3. Tạo ứng dụng tại https://share.streamlit.io và chọn `streamlit_app.py`.
4. Trong ứng dụng, chọn từng trang ở thanh bên; tải file kỳ mới trong mục **Tải báo cáo kỳ mới**.
5. Mở **ROI mục tiêu riêng từng sản phẩm**, nhập các ngưỡng và lưu CSV để dùng lại.
