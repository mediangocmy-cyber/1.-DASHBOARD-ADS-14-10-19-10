## Quy ước chỉ số và ngưỡng cảnh báo

- Không sử dụng ROAS hòa vốn hoặc tỷ suất lợi nhuận.
- Tải lên lần lượt 4 báo cáo theo kỳ mới trong sidebar; mỗi báo cáo tương ứng một trang.
- Các ngưỡng ROI (ROAS), CPA, chi phí và CTR là ngưỡng quản trị có thể điều chỉnh.
